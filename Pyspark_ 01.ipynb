```python
#Q1 Dataset [5, 10, 15, 20, 25, 30] Problem: Create an RDD from the dataset and multiply every number by 2. Expected Output [10, 20, 30, 40, 50, 60]
from pyspark.sql import SparkSession

spark = SparkSession.builder.getOrCreate()
Dataset=[5,10,15,20,25,30]
sc = spark.sparkContext
rdd = sc.parallelize(Dataset)
rdd2= rdd.map(lambda x: x*2)
print(rdd2.collect())
```

```python
# Q22
# Problem:
# Find the number of unique words.
# Expected Output
# 14
from pyspark.sql import SparkSession
spark = SparkSession.builder.getOrCreate()
dataset =["spark is fast",
    "spark is powerful",
    "hadoop is distributed",
    "spark and hadoop are popular",
    "kafka is streaming"]

sc = spark.sparkContext
rdd = sc.parallelize(dataset)
rdd = rdd.flatMap(lambda x: x.split(" "))

# Count occurrences of 'is'
count_is = rdd.filter(lambda word: word == "is").count()

# Get distinct count of non-'is' words
distinct_non_is_words_rdd = rdd.filter(lambda word: word != "is").distinct()
count_distinct_non_is = distinct_non_is_words_rdd.count()

total_unique_count = count_is + count_distinct_non_is
print(total_unique_count)
```

```python
# Q29
# Problem:
# Calculate the average salary for every department.
# Expected Output
# IT -> 88250
# HR -> 62500
# Finance -> 80000
# Sales -> 71000
from pyspark.sql import SparkSession
spark = SparkSession.builder.getOrCreate()
Dataset =[
 ("IT", 80000),
 ("HR", 60000),
 ("IT", 90000),
 ("Finance", 75000),
 ("HR", 65000),
 ("Finance", 85000),
 ("IT", 95000),
 ("Sales", 70000),
 ("Sales", 72000),
 ("IT", 88000)
]
sc = spark.sparkContext
rdd = sc.parallelize(Dataset)
rdd2 = rdd.mapValues(lambda x: (x,1))
rdd3 = rdd2.reduceByKey(lambda x,y: (x[0]+y[0], x[1]+y[1]))
rdd4 = rdd3.mapValues(lambda x: x[0]/x[1])
result = rdd4.collect()
for dept, avg_salary in result:
    print(f"{dept} -> {int(avg_salary)}")
```

```python
# Q50
# Problem:
# Perform the following complete analysis:
# Calculate revenue for every order.
# Keep only orders with revenue greater than 50000.
# Calculate total revenue for each product.
# Find the product with the highest revenue.
# Save the final result into a single output partition.
# Expected Output
# Orders retained:
# O1 -> Laptop -> 100000
# O2 -> Mobile -> 100000
# O6 -> Laptop -> 200000
# Product Revenue:
# Laptop -> 300000
# Mobile -> 100000
# Highest Revenue Product:
# Laptop -> 300000
# Output:
# 1 part file
from pyspark.sql import SparkSession
spark = SparkSession.builder.getOrCreate()

Dataset =[
 ("O1", "Laptop", 2, 50000),
 ("O2", "Mobile", 5, 20000),
 ("O3", "Laptop", 1, 50000),
 ("O4", "Tablet", 3, 15000),
 ("O5", "Mobile", 2, 20000),
 ("O6", "Laptop", 4, 50000)
]
sc = spark.sparkContext
rdd = sc.parallelize(Dataset)
order_revenue_rdd = rdd.map(lambda x: (x[0], x[1], x[2] * x[3]))
filtered_orders = order_revenue_rdd.filter(lambda x: x[2] > 50000).collect()
print("Orders retained:")
for order_id, product_name, revenue in filtered_orders:
    print(f"{order_id} -> {product_name} -> {revenue}")
product_revenue_rdd = rdd.map(lambda x: (x[1], x[2] * x[3]))
total_product_revenue_rdd = product_revenue_rdd.reduceByKey(lambda x, y: x + y)
product_revenues = total_product_revenue_rdd.collect()
print("Product Revenue:")
for product_name, total_revenue in product_revenues:
    print(f"{product_name} -> {total_revenue}")
highest_revenue_product = total_product_revenue_rdd.sortBy(lambda x: x[1], ascending=False).take(1)
print("Highest Revenue Product:")
for product_name, revenue in highest_revenue_product:
    print(f"{product_name} -> {revenue}")


```
