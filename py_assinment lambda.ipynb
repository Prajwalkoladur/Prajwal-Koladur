```python
'''Question 9

Using filter(), map(), and reduce():

Extract temperatures greater than 30°C.
Convert the selected values into Fahrenheit.
Calculate the average Fahrenheit temperature.'''
from functools import reduce
temperatures = [18, 24, 30, 35, 40, 12, 28, 33, 15, 42]
n = filter(lambda x: x>30,temperatures)
m = map(lambda x:(x*9/5)+32,n)
o = reduce(lambda a,b:a+b,m)/len(m)
print(o)
```

```python
# Question 13
"""Using filter(), map(), and reduce():
Extract active employees.
Increase every active employee's salary by 15%.
Calculate the total updated payroll."""

from functools import reduce
employees = [
    {"id": 101, "name": "Rahul", "salary": 45000, "active": True},
    {"id": 102, "name": "Anil", "salary": 65000, "active": True},
    {"id": 103, "name": "Sneha", "salary": 72000, "active": False},
    {"id": 104, "name": "Kiran", "salary": 82000, "active": True},
    {"id": 105, "name": "Meena", "salary": 55000, "active": True},
    {"id": 106, "name": "Arjun", "salary": 98000, "active": False}
]
active_employees = list(filter(lambda x: x["active"], employees))
updated_salaries_list = list(map(lambda x: (x["name"], x["salary"] * 1.15), active_employees))
print(updated_salaries_list) # Print names and updated salaries as requested
total_updated_payroll = reduce(lambda a, b: a + b, map(lambda x: x[1], updated_salaries_list))
print(total_updated_payroll)
```

```python
'''Question 20

Using filter(), map(), and reduce():

Extract active salespeople from the South region.
Calculate a 5% incentive for each selected salesperson.
Create a list containing salesperson name, sales amount, and incentive.
Calculate the total incentive payable.'''
from functools import reduce

sales_records = [
    {"salesperson": "Rahul", "region": "South", "sales": 45000, "active": True},
    {"salesperson": "Anil", "region": "North", "sales": 65000, "active": True},
    {"salesperson": "Sneha", "region": "South", "sales": 72000, "active": False},
    {"salesperson": "Kiran", "region": "South", "sales": 82000, "active": True},
    {"salesperson": "Meena", "region": "West", "sales": 55000, "active": True},
    {"salesperson": "Arjun", "region": "South", "sales": 98000, "active": True}
]
filtered_south_active_salespeople = list(filter(lambda x: x["region"]=="South" and x["active"], sales_records))
m = list(map(lambda x: (x["salesperson"], x["sales"], x["sales"] * 0.05), filtered_south_active_salespeople))
o = reduce(lambda a,b: a + b, map(lambda x: x["sales"] * 0.05, filtered_south_active_salespeople))
print(m)
print(o)
```
