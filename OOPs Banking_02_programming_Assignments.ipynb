### Q8. Protect a amount with a property for banking case 1; use sample reference `ACC-1008` and explain the result.

**Hints**

1. List the objects, their responsibilities, and their invariants before coding.
2. Choose composition for has-a relationships; use inheritance only for genuine is-a substitution.
3. Keep balances private and expose controlled deposit/withdraw/transfer behavior.

**Edge cases to test:** empty input, invalid types, boundary values, and a realistic failure path.

```python
class Transfer:
    def __init__(self, reference, amount):
        self.reference = reference
        self._amount = amount  # "private" backing field

    @property
    def amount(self):
        return self._amount

    @amount.setter
    def amount(self, value):
        if value <= 0:
            raise ValueError("amount must be positive")
        self._amount = value

t = Transfer("ACC-1008", 100)
t.amount = 250       # goes through the setter's validation
print(t.amount)
```

### Q33. Use polymorphic payment processors for banking case 1; use sample reference `ACC-1033` and explain the result.

```python
class CardPayment:
    def process(self, amount):
        return f"Charged {amount} to card"

class UpiPayment:
    def process(self, amount):
        return f"Paid {amount} via UPI"

def pay(processor, amount):
    return processor.process(amount)  # works for any processor -- polymorphism

for processor in (CardPayment(), UpiPayment()):
    print("ACC-1033", pay(processor, 750))
```

```python
class TransferLine:
    def __init__(self, description, amount):
        self.description = description
        self.amount = amount

class Transfer:
    def __init__(self, reference, lines):
        self.reference = reference
        self.lines = lines  # Transfer "has-a" list of TransferLine (composition)

    def total(self):
        return sum(line.amount for line in self.lines)

t = Transfer("ACC-1034", [TransferLine("principal", 1000), TransferLine("fee", 25)])
print(t.total())
```

### Q100. Refactor a god object using solid for banking case 3; use sample reference `ACC-1100` and explain the result.

**Hints**

1. List the objects, their responsibilities, and their invariants before coding.
2. Choose composition for has-a relationships; use inheritance only for genuine is-a substitution.
3. Keep balances private and expose controlled deposit/withdraw/transfer behavior.

**Edge cases to test:** empty input, invalid types, boundary values, and a realistic failure path.

```python
# Before: one "god object" doing everything (validation + storage + notification)
# After: split into small classes, each with a single responsibility (SRP from SOLID)

class Validator:
    def validate(self, amount):
        return amount > 0

class Storage:
    def save(self, reference, amount):
        print(f"saved {reference}: {amount}")

class Notifier:
    def notify(self, reference):
        print(f"notified customer about {reference}")

class TransferService:
    def __init__(self, validator, storage, notifier):
        self.validator = validator
        self.storage = storage
        self.notifier = notifier

    def transfer(self, reference, amount):
        if not self.validator.validate(amount):
            raise ValueError("invalid amount")
        self.storage.save(reference, amount)
        self.notifier.notify(reference)

service = TransferService(Validator(), Storage(), Notifier())
service.transfer("ACC-1100", 500)
# Explanation: each class now has one job, making the system easier to test and change
# than a single class doing everything.
```
