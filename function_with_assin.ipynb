```python
'''Assignment 13: Factorial Using a Function
Write a function named calculate_factorial().'''

def calculate_factorial(number):
    if number < 0:
      return "Error: Factorial is not defined for negative numbers."
    elif number == 0 or number == 1:
      return 1
    else:
      factorial = 1
      for i in range(1, number + 1):
        factorial *= i
      return factorial
try:
  number = int(input("Enter a number:"))
  result = calculate_factorial(number)
  print(f"Factorial:{result}")
except ValueError:
  print("Invalid input. Please enter a valid integer.")
```

```python
'''Assignment 16: Reverse a String
Write a function named reverse_string().

Requirements
Accept a string.
Return its reversed version.
Solve it first using slicing.
Write another function that reverses it using a loop.
Example Input
Enter text: Python
Expected Output
Reversed text: nohtyP'''
def reverse_string_slicing(string):
  return string[::-1]

def reverse_string_loop(string):
  reversed_str = ""
  for char in string:
    reversed_str = char + reversed_str
  return reversed_str
text = input("Enter text: ")
reversed_text_slicing = reverse_string_slicing(text)
print(f"Reversed text (slicing): {reversed_text_slicing}")
reversed_text_loop = reverse_string_loop(text)
print(f"Reversed text (loop): {reversed_text_loop}")
```

```python
'''Assignment 17: Palindrome Checker
Write a function named is_palindrome().'''

def is_palindrame(string):
  cleaned_string = "".join(char.lower() for char in string if char.isalnum())
  return cleaned_string == cleaned_string[::-1]
text = input("Enter text: ")
if is_palindrame(text):
  print("The text is a palindrome.")
else:
  print("The text is not a palindrome.")
```

```python
'''Assignment 19: Find the Largest Element in a List.

Requirements:
- Accept a list of numbers.
- Do not use max().
- Return the largest list element.
- Handle an empty list by returning an appropriate message.
'''
def find_list_maximum(numbers):
  if not numbers:
    return "Error: The list is empty."

  maximum = numbers[0]
  for num in numbers:
    if num > maximum:
      maximum = num
  return maximum

try:
  numbers_str = input("Enter numbers (space-separated): ")
  if not numbers_str.strip():
    numbers_list = []
  else:
    numbers_list = [float(x) for x in numbers_str.split()]

  largest_element = find_list_maximum(numbers_list)
  if isinstance(largest_element, str):
    print(largest_element)
  else:
    print(f"The largest element in the list is: {largest_element}")
except ValueError:
  print("Invalid input. Please enter space-separated numbers.")
```

```python
'''Assignment 31: Return Multiple Values
Write a function named number_statistics().

Requirements
Accept a list of numbers.
Return minimum, maximum, total, and average.
Use tuple unpacking to store returned values.
You may use built-in functions in this assignment.'''
def number_statistics(*nums):
  if not nums:

    return None, None, 0, 0
  total = __builtins__.sum(nums)
  minimum = __builtins__.min(nums)
  maximum = __builtins__.max(nums)
  average = total / len(nums)

  return minimum, maximum, total, average

numbers=[10,20,30,30,40]

min_val, max_val, total_val, avg_val = number_statistics(*numbers)
if min_val is not None:
  print(f"Minimum: {min_val}")
  print(f"Maximum: {max_val}")
  print(f"Total: {total_val}")
  print(f"Average: {avg_val}")
else:
  print("The list was empty, no statistics to display.")
```

```python
'''Assignment 32: Function Returning a Dictionary
Write a function named analyze_text().

Requirements
Accept a string.
Return a dictionary containing:
Total characters
Total words
Total vowels
Total digits
Total spaces
Display every result clearly.'''

def analyze_text(text):
  vowels = "aeiouAEIOU"

  word_count = len(text.split())
  vowels_count = __builtins__.sum(1 for ch in text if ch in vowels)
  digits_count = __builtins__.sum(1 for ch in text if ch.isdigit())
  space_count = text.count(" ")
  char_count = len(text)


  return {
      "Total characters": char_count,
      "Total words": word_count,
      "Total vowels": vowels_count,
      "Total digits": digits_count,
      "Total spaces": space_count
  }

sentence = "hello everyone I am Prajwal 23 years old"

analysis_results = analyze_text(sentence)


print("Text Analysis Results:")
for key, value in analysis_results.items():
  print(f"{key}: {value}")
```

```python
'''Assignment 34: Prime Number Function
Write a function named is_prime().

Requirements
Accept one integer.
Return True when the number is prime.
Return False otherwise.
Use the function to print all prime numbers from 1 to 100.
Hint
A prime number has exactly two factors.'''

import math

def is_prime(number):
  if number <= 1:
    return False
  for i in range(2, int(math.sqrt(number)) + 1):
    if number % i == 0:
      return False
  return True

print("Prime numbers from 1 to 100:")
for num in range(1, 101):
  if is_prime(num):
    print(num, end=" ")
print()
```
